```python
import pandas as pd
```

```python
df = pd.read_excel(r"V:\DEVELOPMENT_AID\__PROCESSING\checklist_r\2_functions\test_crstemplate_LT.xlsx", header=1)
df = df[1:].reset_index(drop=True)
```

```python
def check_crsid(df):
    controller=pd.DataFrame(df['CRS Identification N°'].value_counts()).reset_index(drop=False)
    controller.columns = ['CRS Identification N°', 'count']
    crs_id = []
    for i, row in controller.iterrows():
        if row["count"] > 1:
            crs_id.append(row['CRS Identification N°'])
    df.loc[df['CRS Identification N°'].isin(crs_id), 'errormessage'] += "Duplicate CRS ID; "
    return df
```

```python
check_crsid(df)
```
