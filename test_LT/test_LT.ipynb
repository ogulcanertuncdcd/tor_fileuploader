```python
df["errormessage"]= "" #Initialize the error message column
df["warningmessage"]= "" #Initialize the error message column

for i, row in df.iterrows():
    if ( row["Type of flow (main DAC 1 category)"]==10 or row["Type of flow (main DAC 1 category)"]==21 or row["Type of flow (main DAC 1 category)"]==60 ) & (pd.isna(row["Co-operation modality (replaces the type of aid)"])):
        df.loc[i,"errormessage"]= df.loc[i,"errormessage"]+"; ""Co-operation modality (replaces the type of aid) is missing for this type of flow"
        
    if ( row["Type of flow (main DAC 1 category)"]==10 or row["Type of flow (main DAC 1 category)"]==21 or row["Type of flow (main DAC 1 category)"]==60 ) & (row["Co-operation modality (replaces the type of aid)"]=="A01") & (~row["Sector / Purpose code and corresponding shares"].str.contains("51010")):
        df.loc[i,"errormessage"]= df.loc[i,"warningmessage"]+"; ""Purpose code should be 51010 when co-operation modality is A01"

```
