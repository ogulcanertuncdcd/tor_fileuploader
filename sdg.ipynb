```python
def check_sdg(df):
    import re

    df["SDG focus*"] = df["SDG focus*"].astype(str)

    def check_row(row):
        errormessage = row['errormessage']
        sdg_focus = row['SDG focus*']
        

        # Check for incorrect formatting

        if re.search(r'[^\\d;. ]', sdg_focus):
    
        # if re.search(r'[^\\d;.abcd ]', sdg_focus):
        #     errormessage += "Incorrect formatting of SDGs - please verify that all SDGs are written in numeric format and separated by a semi-colon; "
    
        # Check for duplicate SDGs
        if sdg_focus.str.contains(';').any():
            sdgs = sdg_focus.split(';')

            if len(sdgs) != len(set(sdgs)):
                errormessage += "Duplicate SDG focus - each SDG can only be reported once per activity; "

            if set(sdgs).issubset(list_sdgs):
                pass
            else : 
                errormessage += "SDG focus not in list of valid SDGs; "

                    # Check for more than 10 SDGs
            if sdg_focus.count(';') > 9:
                 errormessage += "A maximum of 10 SDG focuses can be reported per activity; "

        return errormessage

    df['errormessage'] = df.apply(check_row, axis=1)
    return df

```
